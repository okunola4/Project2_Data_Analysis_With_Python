### Q1. Do customers from the UK spend more per transaction than customers from Germany?

```python
Germany = customers_transactions[customers_transactions["country"] == 'Germany']["purchase_value"]
UK = customers_transactions[customers_transactions["country"] == 'UK']["purchase_value"]

t_stat, p_value = ttest_ind(Germany, UK)

print(f"t-statistic: {t_stat:.3f}")
print(f"p-value: {p_value:.5f}")
```

    H0 : Their is no significant difference in mean purchase value of Germany and UK
    
    H1 : Their is a significant difference in mean purchase value of Germany and UK
    
    Since p-value of 0.91082 is greater than 0.05(alpha value)
    
    Conclusion: we accept the null Hypothesis(H0) that their is no significant difference in mean purchase value of Germany     and UK
    
    Meaning that: customers from UK are not spending more than customers from Germany

### Q5. Is device type associated with conversion?

```python
device_type_by_converted = pd.crosstab(web_sessions["device_type"], web_sessions["converted"])
device_type_by_converted
```

```python
chi2, p, dof, expected = chi2_contingency(device_type_by_converted)

print(f"Chi-square statistic: {chi2:.3f}")
print(f"P-value: {p:.5f}")
```

    H0 : Their is no significant relationship(association) between device type and conversion
    
    H1 : Their is a significant relationship(association) between device type and conversion
    
    Since p-value of 0.56265 is greater than 0.05(alpha value)
    
    Conclusion: we accept the null Hypothesis(H0) that their is no significant relationship(association) between 
    device type and conversion

    Meaning: change in device type do not bring about any significant change in conversion

### Explain why statistical significance ≠ business importance.

```python
traffic_source_by_converted = pd.crosstab(transactions_web_sessions["traffic_source"], transactions_web_sessions["converted"])
traffic_source_by_converted
```

### Explain why statistical significance ≠ business importance.
    
    1. Statitistical significance tells if difference exist, not whether that difference actually matters in the real world.
    
    2. The cost of implementing that change or difference may not be justified in real world.
      
    3. With large sample, a tiny and irrelevant difference can become statistically significant.
    
    4  Using accurate data, a tiny inconsequential effect can bring about low p-value.
